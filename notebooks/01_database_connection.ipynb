```python
query = """
SELECT table_name
FROM information_schema.tables
WHERE table_schema='public'
ORDER BY table_name;
"""

tables = db.execute(text(query))

for table in tables:
    print(table[0])
```

```python
query = """
SELECT COUNT(*)
FROM information_schema.tables
WHERE table_schema='public';
"""

count = db.execute(text(query))

print("Total Tables :", count.scalar())
```
