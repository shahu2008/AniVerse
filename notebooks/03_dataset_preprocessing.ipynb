```python
all_genres = set()

for genres in df["genre"]:
    for genre in genres.split(","):
        all_genres.add(genre.strip())

print(f"Total Unique Genres: {len(all_genres)}")
```

```python
sorted(all_genres)
```

```python
genre_df = pd.DataFrame(
    sorted(all_genres),
    columns=["genre_name"]
)

genre_df.head()
```

```python
genre_df.to_csv(
    "../dataset/clean_genres.csv",
    index=False
)

print("Genre list saved successfully!")
```
