```python
chunk_size = 50000

total_imported = 0

valid_anime_ids = set(anime_df["anime_id"])

for chunk in pd.read_csv(
    "../dataset/rating.csv",
    chunksize=chunk_size
):

    # Remove unrated entries
    chunk = chunk[chunk["rating"] != -1]

    # Keep only ratings whose anime exists
    chunk = chunk[
        chunk["anime_id"].isin(valid_anime_ids)
    ]

    objects = []

    for _, row in chunk.iterrows():

        objects.append(
            DatasetRating(
                dataset_user_id=int(row["user_id"]),
                anime_id=int(row["anime_id"]),
                rating=float(row["rating"])
            )
        )

    db.bulk_save_objects(objects)

    db.commit()

    total_imported += len(objects)

    print(f"Imported {total_imported:,} ratings...")
```
